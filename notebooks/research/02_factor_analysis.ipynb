# Quant AI Research — Notebook 02: Quantitative Factor Analysis & Information Coefficient (IC)

Computes factor definitions (Momentum, Volatility, Value, Quality, Sentiment, Liquidity), IC, ICIR, IC decay, and quantile returns.

```python
import sys, os
sys.path.insert(0, os.path.abspath('../../'))
import pandas as pd
import numpy as np
from src.research.factor_analysis import FactorAnalyzer

dates = pd.date_range('2022-01-01', periods=200, freq='B')
df = pd.DataFrame({
    'date': dates, 'ticker': 'AAPL',
    'close': 150 + np.cumsum(np.random.normal(0.1, 1.5, 200)),
    'volume': np.random.randint(1e6, 5e6, 200),
    'sentiment_score': np.random.normal(0.1, 0.4, 200)
})
df['forward_return'] = df['close'].pct_change(5).shift(-5)
df['predicted_signal'] = np.tanh(df['sentiment_score'] * 0.4 + df['close'].pct_change(10).fillna(0) * 1.5)

fa = FactorAnalyzer(df)
fa.compute_factors()
ic_stats = fa.compute_ic('predicted_signal', 'forward_return')
print('IC Stats:', ic_stats)
```
