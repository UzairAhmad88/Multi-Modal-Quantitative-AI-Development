# Phase 8 Notebook 02: Model Registry & Lifecycle Promotion
Demonstrates registering model artifacts, updating signatures, validating models, and performing controlled promotions (EXPERIMENTAL -> VALIDATED -> PAPER -> ARCHIVED).

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parents[1]))

from src.mlops.models import ModelRegistry

model_reg = ModelRegistry()
entry = model_reg.register_model(
    model_name="TransformerAlphaNet",
    model_type="Transformer",
    version="v1.0.0",
    metrics={"predictive.rmse": 0.012, "trading.sharpe_ratio": 1.85}
)
print("Registered Model:", entry["model_id"]) 
promoted = model_reg.update_status(entry["model_id"], "VALIDATED")
print("Promoted Status:", promoted["status"])
```
